```python
#task no 4:
import numpy as np

marks = np.random.randint(40, 101, size=10)

avg_marks = np.mean(marks)
max_marks = np.max(marks)
min_marks = np.min(marks)

print("Generated Marks:", marks)
print("Average Marks  :", avg_marks)
print("Highest Marks  :", max_marks)
print("Lowest Marks   :", min_marks)
```

```python
#task no 5:

marks = np.array([35, 67, 89, 45, 90, 22, 70, 58])

passed_marks = marks[marks >= 50]
failed_marks = marks[marks < 50]

num_passed = len(passed_marks)
num_failed = len(failed_marks)

print("Passed Marks:", passed_marks)
print("Failed Marks:", failed_marks)
print("Number of Passed Students:", num_passed)
print("Number of Failed Students:", num_failed)
```
